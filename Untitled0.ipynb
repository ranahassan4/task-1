```python
import networkx as nx
import matplotlib.pyplot as plt


triples = [
    ("Password", "protects", "Account"),
    ("Two-Factor Authentication", "adds", "Security Layer"),
    ("Phishing", "targets", "Personal Information"),
    ("Personal Information", "must_be", "Protected"),
    ("Antivirus", "blocks", "Malware"),
    ("Firewall", "filters", "Network Traffic"),
    ("Encryption", "secures", "Data"),
]
```

```python
security_graph = nx.DiGraph()


for subj, rel, obj in triples:
    security_graph.add_edge(subj, obj, label=rel)
```

```python
plt.figure(figsize=(12, 7))

pos = nx.kamada_kawai_layout(security_graph)

nx.draw_networkx(
    security_graph,
    pos,
    with_labels=True,
    node_color='#FFD700',
    node_size=2200,
    font_size=10,
    edge_color='#333333',  # Dark Gray for visibility
    arrows=True,
    arrowsize=20,
    connectionstyle='arc3,rad=0.2'  # Adds curved lines
)

edge_labels = nx.get_edge_attributes(security_graph, 'label')
nx.draw_networkx_edge_labels(
    security_graph,
    pos,
    edge_labels=edge_labels,
    font_color='darkred',
    font_size=9,
    label_pos=0.5,
    rotate=False  # Keeps labels horizontal
)

plt.title("Online Security Knowledge Graph", fontsize=14)
plt.axis('off')
plt.show()
```
